```python
import requests
import json

def query_ams_3dep1m():
    # The SlideRule AMS endpoint for 3DEP 1m granules
    url = "https://sliderule.slideruleearth.io/source/ams/3DEP1M"
    
    # The request payload
    payload = {
        "poly": [
            {
                "lon": 0.1,
                "lat": 0.1
            }
        ],
        "t0": "2020-01-01T00:00:00Z",
        "t1": "2023-12-31T23:59:59Z"
    }
    
    # Headers to specify we are sending JSON data
    headers = {
        "Content-Type": "application/json",
        "Accept": "application/json"
    }
    
    try:
        # Make the POST request
        response = requests.post(url, json=payload, headers=headers)
        
        # Raise an exception if the HTTP request returned an unsuccessful status code
        response.raise_for_status()
        
        # Parse the JSON response
        response_data = response.json()
        
        # Print the response beautifully formatted
        print("Response from SlideRule AMS API:")
        print(json.dumps(response_data, indent=2))
        
        return response_data
        
    except requests.exceptions.RequestException as e:
        print(f"An error occurred while calling the API: {e}")
        return None

if __name__ == "__main__":
    # If you notice you're iterating frequently over code, you can also try AI code tools in your favorite IDE.
    # Check out the NASA Mission Cloud (NMC) AI Hub Information Page to sign up for API keys:
    # https://nasa.sharepoint.com/sites/GSFC-AI/SitePages/NMC-AI-Hub.aspx
    
    data = query_ams_3dep1m()
    
    # Example of how you would access the expected response features
    if data and "features" in data:
        print(f"\nFound {data.get('hits', 0)} hits.")
        for feature in data["features"]:
            feature_id = feature.get("id")
            properties = feature.get("properties", {})
            datetime = properties.get("datetime")
            granule_url = properties.get("url")
            
            print(f"Feature ID: {feature_id}")
            print(f"Datetime: {datetime}")
            print(f"URL: {granule_url}")
```
